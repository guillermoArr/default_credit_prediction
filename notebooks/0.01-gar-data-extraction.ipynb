**Main objective**: Extracting dataset _Default Credit Clients_ from UC Irivine Machine Learning Repository.

```python
# -------------------- Libraries ------------------
## Directories and system
import os 

## Data usage
import pandas as pd
import numpy as np

## Data repository
from ucimlrepo import fetch_ucirepo
```

```python
# -------------------- Directories ------------------
# Get main directory
CURRENT_DIR = os.getcwd()
MAIN_DIR = os.path.dirname(CURRENT_DIR)
# Create data directory
os.makedirs(os.path.join(MAIN_DIR, 'data', 'raw'), exist_ok=True)
```

```python
# fetch dataset
default_of_credit_card_clients = fetch_ucirepo(id=350) 

# Read data
X = default_of_credit_card_clients.data.features
y = default_of_credit_card_clients.data.targets
print(f"Shapes - X: {X.shape}, y: {y.shape}")
```

```python
default_of_credit_card_clients.data.ids
```

Podemos igorar el ID ya que es únicamente de referencia y no se utilizará para el modelo

```python
# Merge into a single dataframe
df = pd.concat([X, y], axis=1)
df.head()
```

```python
# metadata 
print(default_of_credit_card_clients.metadata) 
```

```python
# variable information 
print(default_of_credit_card_clients.variables)
```

```python
vars = default_of_credit_card_clients.variables[['name', 'description']]
vars.set_index('name', inplace=True)
df.rename(columns=vars.to_dict()['description'], inplace=True)
df.head()
```

```python
# -------------------- Data Loading ------------------
df.to_csv(os.path.join(MAIN_DIR, 'data', 'raw', 'credit_dataset.csv'), index=False)
```
